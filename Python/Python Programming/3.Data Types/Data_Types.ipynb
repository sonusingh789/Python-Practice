# Data Types in Python

- Numeric Type
   - int 
   - float
   - complex

```python
x  = 5 
y = 5.0
z = 2 + 3j

print(type(x))
print(type(y))
print(type(z))
```

- String Type (str)

```python
a = "sonu"
print(type(a))
```

- Boolean Type (bool)

```python
is_valid = False
is_true = True

print(type(is_valid))
print(type(is_true))
```

- Built in collection or data stucture data types
   - list
   - dict
   - tuple
   - set
    

list and tuple are Sequence types. 

They represent ordered collections of elements where each item can be accessed by its index (position).

dict (Dictionary) is a Mapping type. It stores data as an unordered or insertion-ordered collection of key: value pairs. Instead of using a numerical index, you access values by querying their specific unique key

```python
list = [1,3,4,536363]
print(type(list))
```

```python
tuple = (1,2,426,36,46)
print(type(tuple))
```

```python
set = {1,2,3,4}

print(type(set))
```

- None Type

```python
result = None
print(type(result))
```
