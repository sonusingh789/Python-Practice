# Operators

- Arithmetic 
- {   + * / // %  **   }

```python
print(5+5) # addition
print(5-5) # substract
print(5*5) # multiply
print(5/5) # divide
print(5//5) # interger division
print(5**2) # raising power
```

- Comparison/Relational Operators

   - returns Boolean value True or False

   - == , >= , <= , != , <, >

```python
a = 5
b = 6 

print(a==b)
print(a!=b)
print(a>5)
print(a<5)
print(a>=b)
print(a<=b)

```

- Logical Operator
   - and , or , not

```python
a = 10 
b = 30

print(a==10 and b==30)
print(a>b or b<a)
print(not (a>5))
```

- Assignment Operator

  = , += , -= , *= , /= , //= ,**=2

```python
x = 10 
print(x)
x += 5
print(x)
x -= 3
print(x)
x /= 4
print(x)
x //=5
print(x)
x **=6
print(x)
```

- Bitwise Operator
  - & , | , ^ ,>> , << , ~ 

```python
a = 5
b = 3

print(a&b)
print(a|b)
print(a^b)
print(a<<b)
print(a>>b)
print(~a)
```
